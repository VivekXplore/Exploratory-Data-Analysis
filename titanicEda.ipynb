```python
#now we will show rln with male survivours 

male_survive=df[(df["status"]=="survived")&(df["Sex"]=="male")]
male_survive["Age"].value_counts().sort_index()
```

```python
male_survive["Agegroup"]=pd.cut(male_survive["Age"],bins=7) #pd.cuts make tbe age grp in 7 interval
male_survive.groupby("Agegroup").size()
```

```python
import seaborn as sns

sns.histplot(
    data=male_survive,
    x="Age",
    bins=10
)
plt.title("Mlae survivors according to age group")
```

```python
# now same thing with female survivour 
female_survive=df[(df["status"]=="survived")&(df["Sex"]=="female")] # we crated a subset of dataframe wher male and surviour are filtered 
female_survive.head()
```

```python
female_survive["Age"].value_counts().sort_index() #count how many peple are ther of those age 
```

```python
female_survive["Agegroup"]=pd.cut(female_survive["Age"],bins=7)
female_survive.groupby("Agegroup").size()

"""* Takes the Age column (continuous numbers)
* Splits it into 7 intervals (bins)
* Creates a new column called Agegroup 


* Groups rows by each Agegroup
* Counts how many rows fall into each group"""
```

```python
sns.histplot(data=female_survive, x="Age", bins=10)
plt.title(" Number of Female survivour according to Age group")
```
