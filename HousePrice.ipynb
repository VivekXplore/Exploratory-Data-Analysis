```python
# we care for price so we will check if any outlier present in house price 

q3=df["price"].quantile(0.75) #q3
q1=df["price"].quantile(0.25) #q1=1
iqr=q3-q1 

#lower boundary 
lower=q1- 1.5*iqr
upper=q3+ 1.5*iqr

#if the iqr dosent falls in these boundaries then the outlier is detect which we will se visually as well

outliers=df[(df["price"]<lower) | (df["price"]>upper)]

if not outliers.empty:
    print(f"outliers detected in price . found {len(outliers)} outliers")
    print("outlier are present when the price dosent fall in range of lower and higher boundary")
else:
    print("no outlier detected ")
```

For a data being skewed, if the skewness falls between -0.5 to 0.5 then we consider data to be Normal(symmetrical)

# Budget-Based Housing Analysis. 
In this section, we will explore what type of houses buyers can typically expect within a budget range of $100k–$300k.  

We will try to answer questions like:  

-How many bedrooms do you get within this price range?  
-Will there be any waterfront nearby the house ??  
 
-Which cities/zicode would be able to provide houses in this price range? 
-what would be the condition of the house ? 


so form this section onward we will analyze and visualize all these questions from buyers prespective. 

```python
#first set the price range of the house 
price_range=df[(df["price"]>=100000) &(df["price"]<=300000)].sort_values(by="price")

#from above we filtered the dataframe based on our price
price_range
```

```python
#now we will  analyze how many bedroom do we get within this price range
price_range["bedrooms"].value_counts()
```

so from above we can see mostly you get 3 bedrooms, you have 503 houses with 3 bedroom in our price range from 100k to 300k. similary you get 2 bedroom, then 4 and so on.

7 bedroom is only in one house. The house price probably should lie in the extreme ends our price range. 

```python
sns.countplot(x="bedrooms",data=price_range)
plt.ylabel("No of houses")
plt.xlabel("No of bedrooms")
plt.title("How many bedroom do you get in price range $100k-$300k?")
plt.tight_layout()
```

we viusalied the thing we analyze above. you mostly get 3 bedroom house the most within the price range.


```python
# Will the house have any waterfront nearby within the price range ??

price_range["waterfront"].value_counts()
```

```python

price_range["waterfront"].value_counts().plot(kind="pie")
plt.title("Do you find any waterfront nearby the house within the price range?")
```

The pie chart shows that waterfront properties are absent (or extremely rare) in the $100k–$300k price range within this dataset, suggesting that waterfront access is strongly associated with higher-priced housing segments.

```python
#In which cities or zipcode do you get house in that price range ? 

price_range["city"].unique()
```

Above are all the cities where you get to live in the price range. 
we will now select the top 10 cities where we get the most house within our specified price range

```python
top_ten_city=price_range["city"].value_counts().head(10)
top_ten_city.sort_values(ascending=True).plot(kind="barh",color="lime")
plt.title("Top 10 Cities (100k–300k Price Range)")
plt.xlabel("Number of Houses")
plt.ylabel("City")
plt.tight_layout()
plt.show()
```

From the analysis, we can conclude seattle has the most houses(neraly 200) that people can buy within the price range of 100k-300k followed kent, auburn

```python
# what is the usual condition of the house in that price range ? 

condition=price_range["condition"].value_counts()
condition.plot(kind="bar")
```

so in the specificed price range, you would mostly get house in the condition '3' parameter.
